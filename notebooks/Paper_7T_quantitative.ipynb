# 7T quantitative model comparison

Identical simulations are evaluated with all models. Seed repetitions are averaged within each subject before reporting mean ± sample SD across the five subjects.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path('..').resolve()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = Path('.').resolve()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

CONFIG = PROJECT_ROOT / 'configs/PaperFig1/model_comparison.yaml'
RUN_EVALUATION = True  # Set to False after the HDF5 result has been calculated.
OVERWRITE = False
LOGARITHMIC_Y = True
FIGSIZE = (8.4, 4.0)
SAVE_PDF = None  # e.g. PROJECT_ROOT / 'SavedGraphics/7T_quantitative.pdf'
```

```python
from walinet.paper_figures.paper_fig1 import run_evaluation
from walinet.paper_figures.paper_fig1.config import load_evaluation_config

evaluation_config = load_evaluation_config(CONFIG)
if RUN_EVALUATION:
    results_path = run_evaluation(CONFIG, overwrite=OVERWRITE)
else:
    results_path = evaluation_config.output_path
    if not results_path.is_file():
        raise FileNotFoundError(
            f'{results_path} does not exist; set RUN_EVALUATION=True.'
        )
results_path
```

```python
from walinet.paper_figures.quantitative_7t import grouped_subject_results

subject_results = grouped_subject_results(results_path)
subject_results
```

```python
from walinet.paper_figures.quantitative_7t import reference_subject_results

reference_by_subject = reference_subject_results(subject_results)
reference_mean = reference_by_subject.mean()
reference_std = reference_by_subject.std(ddof=1)
print('Simulated input reference: ||n||_2 / ||m||_2')
print(f'Mean ± SD across {len(reference_by_subject)} subjects: ' 
      f'{reference_mean:.6g} ± {reference_std:.6g}')
print('\nSubject means:')
for subject, value in reference_by_subject.items():
    print(f'  {subject}: {value:.6g}')
```

```python
from walinet.paper_figures.quantitative_7t import create_quantitative_figure

figure, subject_results = create_quantitative_figure(
    results_path,
    logarithmic=LOGARITHMIC_Y,
    figsize=FIGSIZE,
)
if SAVE_PDF is not None:
    save_path = Path(SAVE_PDF).expanduser()
    save_path.parent.mkdir(parents=True, exist_ok=True)
    figure.savefig(save_path, format='pdf', bbox_inches='tight')
    print(f'Saved: {save_path.resolve()}')

import matplotlib.pyplot as plt
plt.show()
```
