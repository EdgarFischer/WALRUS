# 3T LCModel fit-rate comparison

A successful fit is a brain-mask voxel with a finite CRLB satisfying `0 < CRLB <= threshold`. Fit rates are calculated per subject and summarized as mean ± sample SD across validation subjects.

```python
from pathlib import Path

PROJECT_ROOT = Path('..').resolve()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = Path('.').resolve()

CRLB_THRESHOLD = 20.0  # successful fit: 0 < CRLB <= threshold
METHODS = {
    'HLSVD + L2':     'LCModelHSVDL2',
    'WALRUS':         'LCModelAfterWalinet',
}
METABOLITES = {
    'tNAA': 'NAA_NAAG',
    'tCr':  'Cr_PCr',
    'tCho': 'GPC_PCh',
    'Glx':  'Glu_Gln',
}
FIGSIZE = (7.0, 4.2)
SAVE_PDF = None  # e.g. PROJECT_ROOT / 'results/PaperFigures/3T_fit_rate.pdf'
```

```python
import sys
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

from walinet.paper_figures.lcmodel_quality import collect_fit_rates, summary_table

subject_fit_rates = collect_fit_rates(
    project_root=PROJECT_ROOT,
    crlb_threshold=CRLB_THRESHOLD,
    methods=METHODS,
    metabolites=METABOLITES,
)
summary = summary_table(subject_fit_rates)
summary
```

```python
from walinet.paper_figures.lcmodel_quality import plot_fit_rates

figure = plot_fit_rates(subject_fit_rates, figsize=FIGSIZE)
if SAVE_PDF is not None:
    save_path = Path(SAVE_PDF).expanduser()
    save_path.parent.mkdir(parents=True, exist_ok=True)
    figure.savefig(save_path, format='pdf', bbox_inches='tight')
    print(f'Saved: {save_path.resolve()}')
import matplotlib.pyplot as plt
plt.show()
```
